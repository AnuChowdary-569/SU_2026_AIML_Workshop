```python
#DataVisualization --> matplotlib,seaborn,plotly
import matplotlib.pyplot as plt
```

```python
#Distribution of Age
plt.hist(data['Age'],bins = 15,color = 'pink')
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")
plt.show()
```

```python
#Distribution of Age
counts,edges,bars = plt.hist(data['Age'],bins = 15,color = 'pink')
print(counts)
print(bars)
plt.xlabel("Age")
plt.ylabel("Count")
plt.bar_label(bars)
plt.title("Age Distribution")
plt.show()
```
